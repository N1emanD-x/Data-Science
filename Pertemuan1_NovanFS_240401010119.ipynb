```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches

print("\n📊 Membuat dataset contoh industri Data Science di Indonesia...\n")

# Dataset sederhana: perusahaan & penggunaan Data Science
data = {
    "Perusahaan"     : ["Gojek", "Tokopedia", "Bank BRI", "BPJS Kesehatan", "Traveloka"],
    "Sektor"         : ["Teknologi", "E-Commerce", "Perbankan", "Kesehatan", "Perjalanan"],
    "Use_Case"       : [
        "Prediksi permintaan & fraud detection",
        "Sistem rekomendasi produk",
        "Penilaian kredit mikro UMKM",
        "Deteksi fraud klaim",
        "Dynamic pricing tiket & hotel"
    ],
    "Tim_Data_Approx": [200, 150, 100, 80, 120],   # perkiraan jumlah tim data
    "Maturity_Score" : [9, 8, 7, 6, 8],             # skala 1–10
}

df = pd.DataFrame(data)

print("=== DataFrame Industri Data Science Indonesia ===")
print(df[["Perusahaan", "Sektor", "Tim_Data_Approx", "Maturity_Score"]])

print("\n=== Statistik Deskriptif ===")
print(df[["Tim_Data_Approx", "Maturity_Score"]].describe().round(2))

# --- Visualisasi ---
fig, axes = plt.subplots(1, 2, figsize=(12, 5))
fig.suptitle("Data Science di Industri Indonesia", fontsize=14, fontweight="bold")

# Plot 1: Bar chart ukuran tim
colors_bar = ["#1D9E75", "#3B8BD4", "#EF9F27", "#D85A30", "#534AB7"]
axes[0].barh(df["Perusahaan"], df["Tim_Data_Approx"], color=colors_bar)
axes[0].set_xlabel("Estimasi Jumlah Tim Data")
axes[0].set_title("Ukuran Tim Data per Perusahaan")
axes[0].invert_yaxis()
for i, v in enumerate(df["Tim_Data_Approx"]):
    axes[0].text(v + 2, i, str(v), va="center", fontsize=10)

# Plot 2: Scatter plot maturity vs tim
scatter_colors = colors_bar
sc = axes[1].scatter(
    df["Tim_Data_Approx"],
    df["Maturity_Score"],
    c=colors_bar,
    s=150,
    zorder=5
)
for i, row in df.iterrows():
    axes[1].annotate(
        row["Perusahaan"],
        (row["Tim_Data_Approx"], row["Maturity_Score"]),
        textcoords="offset points",
        xytext=(8, 4),
        fontsize=9
    )
axes[1].set_xlabel("Estimasi Jumlah Tim Data")
axes[1].set_ylabel("Data Maturity Score (1–10)")
axes[1].set_title("Ukuran Tim vs Kematangan Data")
axes[1].set_ylim(4, 11)
axes[1].grid(True, alpha=0.3)

plt.tight_layout()
plt.savefig("visualisasi_industri_ds.png", dpi=120, bbox_inches="tight")
plt.show()
print("\n✅ Grafik tersimpan sebagai 'visualisasi_industri_ds.png'")
```
