```python
# ===== [STEP 1] IMPORT LIBRARY =====
import pandas as pd
import numpy as np
```

```python
# ===== [STEP 2] LOAD DATASET =====
url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
df = pd.read_csv(url)

# Tampilkan 5 data pertama
df.head()
```

```python
# ===== [STEP 3] EKSPLORASI DATA =====

# Ukuran dataset
print("Shape:", df.shape)

# Nama kolom
print("Kolom:", df.columns.tolist())

# Info dataset
print("\nInfo Data:")
df.info()

# Statistik deskriptif
print("\nStatistik:")
df.describe()

# Missing values
print("\nMissing Values:")
df.isnull().sum()
```

```python
# ===== [Q1] JUMLAH BARIS & KOLOM =====
df.shape
```

```python
# ===== [Q2] MISSING VALUES =====
df.isnull().sum()
```

```python
# ===== [Q3] PERSENTASE YANG SELAMAT =====
persentase_selamat = df['Survived'].mean() * 100
persentase_selamat
```

```python
# ===== [STEP 4] FILTERING =====

# Wanita kelas 1
wk1 = df[(df['Sex']=='female') & (df['Pclass']==1)]

# Lihat data
wk1.head()
```

```python
# ===== [Q4] WANITA KELAS 1 =====

jumlah = wk1.shape[0]
rata_usia = wk1['Age'].mean()

print("Jumlah:", jumlah)
print("Rata-rata usia:", rata_usia)
```

```python
# ===== [STEP 5] GROUPBY =====
group = df.groupby('Pclass')['Survived'].mean()
group
```

```python
# ===== [Q5] KELAS DENGAN SURVIVAL TERTINGGI =====
group
```

```python
# ===== [Q6] ANALISIS TAMBAHAN =====

# Survival berdasarkan gender
gender = df.groupby('Sex')['Survived'].mean()
gender
```
